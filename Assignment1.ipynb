```python
# Task 2: Python, NumPy & Pandas
# 2.1 Verify Statistics with NumPy

import numpy as np

marks = np.array([72, 85, 90, 65, 78, 92, 68, 75, 88, 80])

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Range:", np.max(marks) - np.min(marks))
print("Variance (sample):", np.var(marks, ddof=1))
print("Std Deviation (sample):", np.std(marks, ddof=1))
```

```python
# 2.2 Pandas DataFrame Operations

import pandas as pd

# 1. Load data into Pandas DataFrame
data = {
    'emp_id': ['E01', 'E02', 'E03', 'E04', 'E05', 'E06', 'E07', 'E08', 'E09', 'E10'],
    'name': ['Aarav', 'Bhavya', 'Chirag', 'Diya', 'Eshan', 'Farah', 'Gaurav', 'Hina', 'Ishan', 'Jiya'],
    'department': ['IT', 'HR', 'IT', 'Finance', 'IT', 'Sales', 'Finance', 'HR', 'Sales', 'IT'],
    'age': [28, 32, 26, 35, 30, 29, 40, 27, 33, 38],
    'salary': [55000, 48000, 52000, 72000, 60000, 45000, 85000, 42000, 55000, 78000],
    'experience_yrs': [3, 5, 2, 8, 4, 3, 12, 2, 6, 10]
}
df = pd.DataFrame(data)

# 2. Display the first 5 rows
print("--- First 5 Rows ---")
print(df.head())

# 3. Print summary statistics
print("\n--- Summary Statistics ---")
print(df.describe())

# 4. Average salary by department
print("\n--- Average Salary by Department ---")
print(df.groupby('department')['salary'].mean())

# 5. Filter employees with experience > 5 years
print("\n--- Employees with Experience > 5 Years ---")
print(df[df['experience_yrs'] > 5])

# 6. Sort by salary descending and display top 3
print("\n--- Top 3 Highest Earners ---")
print(df.sort_values(by='salary', ascending=False).head(3))

# 7. Add salary_per_year_exp column
df['salary_per_year_exp'] = df['salary'] / df['experience_yrs']
print("\n--- DataFrame with salary_per_year_exp Column ---")
print(df)
```

```python
# 3.3 Matplotlib Visualization

import pandas as pd
import matplotlib.pyplot as plt

# Dataset Setup
data = {
    'emp_id': ['E01', 'E02', 'E03', 'E04', 'E05', 'E06', 'E07', 'E08', 'E09', 'E10'],
    'name': ['Aarav', 'Bhavya', 'Chirag', 'Diya', 'Eshan', 'Farah', 'Gaurav', 'Hina', 'Ishan', 'Jiya'],
    'department': ['IT', 'HR', 'IT', 'Finance', 'IT', 'Sales', 'Finance', 'HR', 'Sales', 'IT'],
    'age': [28, 32, 26, 35, 30, 29, 40, 27, 33, 38],
    'salary': [55000, 48000, 52000, 72000, 60000, 45000, 85000, 42000, 55000, 78000],
    'experience_yrs': [3, 5, 2, 8, 4, 3, 12, 2, 6, 10]
}
df = pd.DataFrame(data)

# 1. Bar Chart: Department vs Average Salary
avg_salary_dept = df.groupby('department')['salary'].mean().reset_index()
plt.figure(figsize=(7, 4))
plt.bar(avg_salary_dept['department'], avg_salary_dept['salary'], color='teal')
plt.title('Department vs Average Salary')
plt.xlabel('Department')
plt.ylabel('Average Salary ($)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# 2. Scatter Plot: Experience vs Salary by Department
plt.figure(figsize=(7, 4))
departments = df['department'].unique()
colors = ['red', 'blue', 'green', 'purple']

for dept, color in zip(departments, colors):
    subset = df[df['department'] == dept]
    plt.scatter(subset['experience_yrs'], subset['salary'], label=dept, color=color, s=80)

plt.title('Experience vs Salary by Department')
plt.xlabel('Experience (Years)')
plt.ylabel('Salary ($)')
plt.legend(title='Department')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# 3. Histogram: Age Distribution (Bins of 5)
plt.figure(figsize=(7, 4))
plt.hist(df['age'], bins=range(25, 46, 5), color='darkorange', edgecolor='black', alpha=0.8)
plt.title('Distribution of Employee Ages')
plt.xlabel('Age (Years)')
plt.ylabel('Frequency')
plt.xticks(range(25, 46, 5))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
